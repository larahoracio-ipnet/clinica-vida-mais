```python
# arquivo json da ia no volume
caminho_json_ia = "/Volumes/workspace/default/dados-ia/lote_12_guias.jsonl" 

#leitura do arquivo
df_bronze_ia = (
    spark.read
    .option("encoding", "UTF-8")
    .json(caminho_json_ia)
)

#salvando o arquivo em uma delta table
df_bronze_ia.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("bronze_ia")
display(spark.table("bronze_ia"))
```
