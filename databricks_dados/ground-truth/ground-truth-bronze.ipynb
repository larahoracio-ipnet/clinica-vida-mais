```python
from pyspark.sql import functions as F
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
```

```python
#Criacao de um Schema rígido (protege zeros à esquerda nos códigos TUSS)
schema_gabarito = StructType([
    StructField("id_guia", StringType(), False),
    StructField("item", IntegerType(), False),
    StructField("nome", StringType(), True),
    StructField("carteirinha", StringType(), True),
    StructField("codigo_tuss", StringType(), True),
    StructField("descricao", StringType(), True)
])
```

```python
#leiitura do arquivo csv que possui os dados das guias hospedado no volume
dados_gabarito = spark.read.csv("/Volumes/workspace/default/csv-bronze/gabarito.csv", header=True, schema=schema_gabarito,sep=";")

#apaga a tabela anterior caso ja exista
spark.sql("DROP TABLE IF EXISTS gabarito")

#Criacao da tabela
dados_gabarito.write.mode("overwrite").saveAsTable("gabarito")
```

```python
#Leitura do gabarito
df_leitura = spark.table("gabarito")
display(df_leitura)
```
